```python
#####spaCy############
import spacy

nlp = spacy.load("en_core_web_lg")
doc = nlp("I live on 123 Main Street and work near 5th Avenue.")
ner_matches = [ent.text for ent in doc.ents if ent.label_ in {"LOC", "GPE", "FAC"}]
print(ner_matches)
doc = nlp("I need a route from haneviim 37")
ner_matches = [ent.text for ent in doc.ents if ent.label_ in {"LOC", "GPE", "FAC"}]
print(ner_matches)
```

```python
#####Hugging Face Transformers############
from transformers import pipeline

ner_pipeline = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english")
text = "I live on 123 Main Street and work near 5th Avenue."
ner_results = ner_pipeline(text)
print(ner_results)
text = "I need a route from haneviim 37"
ner_results = ner_pipeline(text)
print(ner_results)
```

```python
######flair############
from flair.models import SequenceTagger
from flair.data import Sentence

tagger = SequenceTagger.load("flair/ner-english")
sentence = Sentence("I live on 123 Main Street and work near 5th Avenue.")
tagger.predict(sentence)
print(sentence.to_dict(tag_type="ner"))
sentence = Sentence("I need a route from haneviim 37")
tagger.predict(sentence)
print(sentence.to_dict(tag_type="ner"))
```

```python
###########Stanford NLP (Stanza)
import stanza

stanza.download("en")
nlp = stanza.Pipeline("en")
print()
doc = nlp("I live on 123 Main Street and work near 5th Avenue.")
for ent in doc.entities:
    print(ent.text, ent.type)

print()

doc = nlp("I need a route from haneviim 37")
for ent in doc.entities:
    print(ent.text, ent.type)
```
