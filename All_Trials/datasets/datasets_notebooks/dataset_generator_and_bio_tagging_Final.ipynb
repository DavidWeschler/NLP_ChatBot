```python
sentences = [sentence.replace(" {}", "") for sentence in sentences]
import re

def extract_entities(sentence):
    """
    Extract entities from the sentence, e.g., {difficulty: easy}.
    Returns a list of tuples containing (entity_type, entity_value) and their positions in the sentence.
    """
    matches = [(match.start(), match.end(), *re.match(r'\{(\w+): ([^\}]+)\}', match.group()).groups())
               for match in re.finditer(r'\{(\w+): [^\}]+\}', sentence)]
    return matches

def tokenize_with_entities(sentence):
    """
    Tokenize the sentence while handling entities and generating BIO tags.
    """
    tokens = []
    bio_tags = []
    current_index = 0

    # Extract entities with positions
    entities = extract_entities(sentence)

    # Iterate over entities and plain text
    for start, end, entity_type, entity_value in entities:
        # Process plain text before the entity
        plain_text = sentence[current_index:start].strip()
        if plain_text:
            plain_tokens = plain_text.split()
            tokens.extend(plain_tokens)
            bio_tags.extend(['O'] * len(plain_tokens))

        # Process entity tokens
        entity_tokens = entity_value.split()
        tokens.extend(entity_tokens)
        bio_tags.append(f"B-{entity_type}")
        bio_tags.extend([f"I-{entity_type}"] * (len(entity_tokens) - 1))

        # Update current index
        current_index = end

    # Process remaining plain text after the last entity
    plain_text = sentence[current_index:].strip()
    if plain_text:
        plain_tokens = plain_text.split()
        tokens.extend(plain_tokens)
        bio_tags.extend(['O'] * len(plain_tokens))

    return tokens, bio_tags

def process_sentences(arr):
    result = []

    for sentence in arr:
        sentence = sentence.strip()
        tokens, bio_tags = tokenize_with_entities(sentence)
        result.append({'sentence': sentence, 'tokens': tokens, 'bio_tags': bio_tags})

    return result

# Process the sentences
processed_sentences = process_sentences(sentences)

# Display results
for i, obj in enumerate(processed_sentences, start=1):
  if i == 10:
    break
  print(f"{i}. sentence: {obj['sentence']}")
  print(f"tokens: {obj['tokens']}")
  print(f"bio_tags: {obj['bio_tags']}")
  print("------------------------------------------------")

print("*"*100)
print("                                            DONE!")
print("*"*100)
```
